```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)
```

```python
num_samples = 100
true_weight = 4.5
true_bias = 2.0
```

```python
x = np.random.uniform(-10, 10, size=(num_samples, 1))
noise = np.random.normal(loc=0.0, scale=9.0, size = (num_samples, 1))
y = true_weight * x + true_bias + noise
```

```python

plt.scatter(x,y)
```

```python
x.dtype
```

```python
mean_x = x.mean()
mean_y = y.mean()
```

```python
def cal_slope(mean_x: int, mean_y: int):
    num = 0
    den = 0
    for i in range(num_samples):
        num += (x[i]-mean_x)*(y[i] - mean_y)
        den += (x[i]-mean_x)**2
    return num/den
```

```python
def cal_intercept(beta_1,mean_x, mean_y):
    beta_0 = mean_y - beta_1*mean_x
    return beta_0
```

```python
beta_1 = cal_slope(mean_x, mean_y)
beta_0 = cal_intercept(beta_1, mean_x, mean_y)
```

## Here z is $\hat{y}$ and we can give different values of x for finding the prediction

```python
z = beta_0 + beta_1*x 
def prediction(x):
    return beta_0 + beta_1*x
```

We have const function as $J(w,b) = 1/{2m} \sum_{i=1}^{m} (f_{w,b}(x_i) - y_i)^2$ here ${y_i}$ is the real value and ${f}$ is the predicted one

```python
def cost_fn(x_train: ndarray,y_train: ndarray, sample_size)->np.float64:
    j = 0
    for i in range(sample_size):
        j += (prediction(x_train[i]) - y_train[i])**2
    final_cost = (1 / (2 * sample_size)) * j
```

```python
def cost(x: np.ndarray, y: np.ndarray) -> np.ndarray:
    sample_size = num_samples
    return (1 / (2 * sample_size)) * (prediction(x) - y) ** 2
```
